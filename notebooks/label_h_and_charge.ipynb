# Label check — implicit H and charges

Visual check for the Rust paint fixes:

1. **Implicit H** on N/O must appear (`OH`/`HO`, `NH₂`/`H₂N`), not bare `O`/`N`.
2. **Charges** (`+` / `−`) should be large enough to read and clear of bond ends.
3. **Tetravalent N⁺** (four bonds + formal charge) is the hard bond-overlap case —
   the `+` must not sit on any of the four stems.

````python
from xpict import mol
````

## Implicit H on oxygen and nitrogen

Ethanol, phenol, aniline, ethylamine — look for `OH`/`HO` and `NH₂`/`H₂N`.


````python
(
    mol("CCO", label="ethanol")
    | mol("c1ccccc1O", label="phenol")
    | mol("c1ccccc1N", label="aniline")
    | mol("CCN", label="ethylamine")
)
````

## Charges (size)

Ammonium, hydroxide, acetate, guanidinium — `+`/`−` should be readable.


````python
(
    mol("[NH4+]", label="ammonium")
    | mol("[OH-]", label="hydroxide")
    | mol("CC(=O)[O-]", label="acetate")
    | mol("NC(=[NH2+])N", label="guanidinium")
)
````

## Tetravalent N⁺ — bond overlap stress test

Quaternary ammoniums put **four bonds** into the charged nitrogen. Zoom the
SVG: the `N⁺` (or `⁺` alone next to N) must clear every stem — especially
diagonal bonds that used to clip the superscript.


````python
(
    mol("C[N+](C)(C)C", label="Me4N+")
    | mol("CC[N+](CC)(CC)CC", label="Et4N+")
    | mol("c1ccc(C[N+](C)(C)C)cc1", label="BnMe3N+")
    | mol("C[N+]1(C)CCCCC1", label="Me2piperidinium")
)
````

### Larger / asymmetric tetravalent N⁺

Crowded or off-center charge placement makes overlap more obvious.


````python
(
    mol("C[N+](C)(C)CC", label="EtMe3N+")
    | mol("C1CC[N+](C)(C)CC1", label="Me2piperidinium-alt")
    | mol("c1ccc([N+](C)(C)C)cc1", label="PhMe3N+")
    | mol("C[N+](C)(C)CCO", label="choline-like")
)
````

## Side-by-side: charged vs neutral heteroatoms

Same scaffolds with and without formal charge.


````python
(
    (mol("CCO", label="EtOH") + mol("CC[O-]", label="EtO−"))
    | (mol("CCN", label="EtNH2") + mol("CC[NH3+]", label="EtNH3+"))
    | (mol("c1ccccc1O", label="PhOH") + mol("c1ccccc1[O-]", label="PhO−"))
)
````

## Optional: label ink boxes

`debug_box=True` draws dashed AABBs around label ink — useful if a charge
still looks tight to a bond. Tetravalent N⁺ is the clearest case.


````python
(
    mol("C[N+](C)(C)C").opts(debug_box=True)
    | mol("c1ccc(C[N+](C)(C)C)cc1").opts(debug_box=True)
    | mol("CC(=O)[O-]").opts(debug_box=True)
)
````
