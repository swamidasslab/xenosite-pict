# Default molecule align on larger mols and macrocycles

`mol(a) + mol(b)` uses house MCS (100 ms / 50k states) then CoordGen **soft-constrain**.
This notebook is the visual check: correspondence can look fine while the pose does not.

Run `python notebooks/align_large_pairs.py` first (writes `artifacts/align_large/*.png`), or execute the cell below to regenerate.

```python
import sys
from pathlib import Path
from IPython.display import Image, Markdown, display

nbdir = Path.cwd() / "notebooks" if (Path.cwd() / "notebooks").is_dir() else Path.cwd()
sys.path.insert(0, str(nbdir))
from align_large_pairs import OUT, PAIRS, run_pair

if not (OUT / "stats.json").exists():
    for key, n1, s1, n2, s2 in PAIRS:
        run_pair(key, n1, s1, n2, s2, mode=None)
        run_pair(key, n1, s1, n2, s2, mode="skip")

rows = [
    ("morphine_codeine", "Fused pentacycle. Skip already matches; default rotates codeine."),
    ("cholesterol_testosterone", "Steroid. Default locks testosterone onto a bad cholesterol template."),
    ("crown6_aza", "18-membered crown. Default pose lock is tight."),
    ("crown6_dibenzo", "Same ring, extra benzo. Macrocycle ovalizes."),
    ("crown4_crown5", "12 vs 15 crown. Size mismatch; query collapses."),
    ("c12_c15", "Aliphatic rings 12 vs 15. Query tangles."),
    ("erythro_clarithro", "14-membered macrolides, 6-OH vs 6-OMe. Partial overlay."),
    ("erythro_azithro", "14 vs 15 macrolide (ring expansion). Azithromycin spaghetti."),
]
for key, note in rows:
    display(Markdown(f"## {key}\n{note}\n\n**default** then **skip**"))
    for mode in ("default", "skip"):
        p = OUT / f"{key}_{mode}.png"
        if p.exists():
            display(Image(str(p)))
```
